```python
# helper to run one sweep and return a DataFrame
def sweep_and_collect(param_name, values, **fixed_kwargs):
    records = []
    for v in values:
        # build kwargs for this iteration
        kwargs = {**fixed_kwargs,
                  param_name: v}
        exemplars, centroids, labels = cat.make_categories(**kwargs)
        records.append({
            param_name: v,
            "exemplars": exemplars,
            "centroids": centroids
        })
    return pd.DataFrame(records)
```

```python
# Generate MDS and t-SNE for one row and save the figure.
# The function now takes separate axis-limit tuples for MDS and t-SNE.
def generate_mds_tsne_for_row(
    row_data, row_index, param_name, param_value,
    mds_limits=None, tsne_limits=None
):
    """
    Plot MDS and t-SNE for the exemplars in one sweep row and
    save the figure.  Axis limits can be passed independently:
        mds_limits  = (x_min, x_max, y_min, y_max)
        tsne_limits = (x_min, x_max, y_min, y_max)
    """
    exemplars = row_data["exemplars"]
    print(f"Processing {param_name}={param_value}  (row {row_index})")

    # -------------------- distance matrix --------------------
    distMat = dist.squareform(dist.pdist(exemplars, metric="euclidean"))

    # -------------------- figure --------------------
    fig, axs = plt.subplots(1, 2, figsize=(14, 7))
    fig.suptitle(f"{param_name} = {param_value}", fontsize=16)

    # -------------------- 1) MDS --------------------
    mds = MDS(n_components=2, dissimilarity="precomputed", random_state=42)
    mds_xy = mds.fit_transform(distMat)

    colors = sns.color_palette(n_colors=4)
    for i in range(4):
        s, e = i * 200, (i + 1) * 200     # 200 exemplars per category
        axs[0].scatter(
            mds_xy[s:e, 0], mds_xy[s:e, 1],
            s=20, alpha=0.6, color=colors[i], label=f"Category {i}"
        )
    axs[0].set_title("MDS – Exemplars")
    axs[0].legend()
    if mds_limits is not None:
        x_min, x_max, y_min, y_max = mds_limits
        axs[0].set_xlim(x_min, x_max)
        axs[0].set_ylim(y_min, y_max)

    # -------------------- 2) t-SNE --------------------
    tsne = TSNE(n_components=2, metric="precomputed",
                init="random", random_state=42)
    tsne_xy = tsne.fit_transform(distMat)

    for i in range(4):
        s, e = i * 200, (i + 1) * 200
        axs[1].scatter(
            tsne_xy[s:e, 0], tsne_xy[s:e, 1],
            s=20, alpha=0.6, color=colors[i], label=f"Category {i}"
        )
    axs[1].set_title("t-SNE – Exemplars")
    axs[1].legend()
    if tsne_limits is not None:
        x_min, x_max, y_min, y_max = tsne_limits
        axs[1].set_xlim(x_min, x_max)
        axs[1].set_ylim(y_min, y_max)

    # -------------------- finish --------------------
    plt.tight_layout()
    filename = f"mds_tsne_{param_name}_{param_value}.png"
    plt.savefig(filename, dpi=300, bbox_inches="tight", facecolor="white")
    print(f"Saved → {filename}")
    plt.show()

    return mds_xy, tsne_xy
```

```python
# --------------------------------------------
# 0)  which sweeps do you want to visualise?
#     (add / remove pairs as you like)
# --------------------------------------------
dataframes = [
    (features_df, 'nFeatures'),
    (sub_df,      'sub_rad'),
]

# --------------------------------------------
# 1)  compute global limits SEPARATELY for MDS and t-SNE
# --------------------------------------------
mds_x_min, mds_x_max =  np.inf, -np.inf
mds_y_min, mds_y_max =  np.inf, -np.inf
tsn_x_min, tsn_x_max =  np.inf, -np.inf
tsn_y_min, tsn_y_max =  np.inf, -np.inf

for df, _ in dataframes:
    for _, row in df.iterrows():
        X = row['exemplars']
        D = dist.squareform(dist.pdist(X, metric='euclidean'))

        # -- MDS --
        mds_xy = MDS(n_components=2, dissimilarity='precomputed',
                     random_state=42).fit_transform(D)
        mds_x_min, mds_x_max = min(mds_x_min, mds_xy[:,0].min()), max(mds_x_max, mds_xy[:,0].max())
        mds_y_min, mds_y_max = min(mds_y_min, mds_xy[:,1].min()), max(mds_y_max, mds_xy[:,1].max())

        # -- t-SNE --
        tsn_xy = TSNE(n_components=2, metric='precomputed',
                      init='random', random_state=42).fit_transform(D)
        tsn_x_min, tsn_x_max = min(tsn_x_min, tsn_xy[:,0].min()), max(tsn_x_max, tsn_xy[:,0].max())
        tsn_y_min, tsn_y_max = min(tsn_y_min, tsn_xy[:,1].min()), max(tsn_y_max, tsn_xy[:,1].max())

# add 5 % padding
pad = 0.05
def _limits(lo, hi):
    span = hi - lo
    return lo - pad*span, hi + pad*span

mds_limits  = _limits(mds_x_min, mds_x_max) + _limits(mds_y_min, mds_y_max)   # (x_min,x_max,y_min,y_max)
tsne_limits = _limits(tsn_x_min, tsn_x_max) + _limits(tsn_y_min, tsn_y_max)

# --------------------------------------------
# 2)  CALL THE PLOT FUNCTION for every row
# --------------------------------------------
for df, pname in dataframes:
    for idx, row in df.iterrows():
        generate_mds_tsne_for_row(row, idx, pname, row[pname],
                                  mds_limits=mds_limits,
                                  tsne_limits=tsne_limits)
```

```python
def run_ebrw_for_row(row, param_name, c_value=1.0, b_value=0.25, file_prefix="ebrw_results"):
    # Compose ebrw_kwargs
    ebrw_kwargs = {"c": c_value, "b": b_value}
    # Compose file name
    file_name = f"{file_prefix}_{param_name}_{row[param_name]}_c_{c_value}_b_{b_value}.csv"
    # Run simulation
    simReps = row['exemplars']
    performance, accSummary, RTSummary = cat.run_ebrw_simulation(
        ebrw_kwargs=ebrw_kwargs,
        simReps=simReps,
        file_name=file_name
    )
    print(f"Results for {param_name}={row[param_name]}, c={c_value}, b={b_value}:")
    print(accSummary)
    print(RTSummary)
    return performance, accSummary, RTSummary
```

```python
best_nfeatures_indices = [5, 6, 7, 8, 9]
best_sub_rad_indices = [1, 3, 5, 7, 9]
```

```python
def collect_ebrw_results():
    results = []
    param_sweeps = [
        ('nFeatures', features_df, best_nfeatures_indices),
        ('sub_rad', sub_df, best_sub_rad_indices)
    ]
    for param_name, df, best_indices in param_sweeps:
        for i in range(8):
            c_value = i
            for idx in best_indices:
                row = df.iloc[idx]
                _, accSummary, RTSummary = run_ebrw_for_row(row, param_name=param_name, c_value=c_value)
                for level in accSummary.index:
                    results.append({
                        'param': param_name,
                        'param_value': row[param_name],
                        'c': c_value,
                        'level': level,
                        'mean_acc': accSummary.loc[level, 'mean'],
                        'std_acc': accSummary.loc[level, 'std'],
                        'mean_rt': RTSummary.loc[level, 'mean'],
                        'std_rt': RTSummary.loc[level, 'std'],
                    })
    results_df = pd.DataFrame(results)
    return results_df
```

```python
results_df = collect_ebrw_results()
print(results_df.head())
```
